## Energy Difference after one Flip

The energy of the 1D Ising model or chain with nearest neighbour interactions is given by 
$$
E=-\frac{1}{2}\sum_{i=1}^N J(s_is_{i-1}+s_is_{i+1}),
$$
where $N$ is the size of the model and $s_i$ represent normalized spins that can be only $+1$ and $-1$. With open boundary conditions (i.e. $s_1$ and $s_N$ only have one neighbour), the energy reduces to
$$
E=-J\sum_{i=1}^{N-1}s_is_{i+1}.
$$

A spin flip at index $i$ only affects the two interaction terms involving $s_i$ in the sum: 
$$
E = -J\left(... + s_{i-1} s_i + s_i s_{i+1} + ...\right).
$$ 
The energy change is defined as $\Delta E = E_\text{final} - E_\text{initial}$. Since all unaffected terms cancel, this becomes 
$$
\Delta E = -J \left(s_{i-1} s_i^\text{final} + s_i^\text{final} s_{i+1} - s_{i-1} s_i^\text{initial} - s_i^\text{initial} s_{i+1}\right) = -J \left(s_i^\text{final} - s_i^\text{intial}\right)\left(s_{i-1} + s_{i+1}\right).
$$ 
With $s_i^\text{final} = -s_i^\text{initial}$, we obtain 
$$
\boxed{
\Delta E = 2J s_i^\text{intitial} \left(s_{i-1} + s_{i+1}\right)
}
$$

## Heat Capacity per Spin

### General (used in Monte Carlo)

The thermodynamic definition of the heat capacity at constant volume is 
$$
C_V = \left(\frac{\partial\langle E\rangle}{\partial T}\right)_V = \frac{\partial \langle E \rangle}{\partial \beta} \frac{\partial \beta}{\partial T} = -\frac{1}{k_B T^2} \frac{\partial \langle E \rangle}{\partial \beta}.
$$
The mean energy is defined as 
$$
\langle E \rangle = \sum_s E_s P_s,
$$
where $P_s$ is the probability of finding the system in a configuration with energy $E_s$ and is given by the Boltzmann distribution 
$$
P_s = \frac{e^{-\beta E_s}}{Z}.
$$
$Z$ is the canonical partition function defined as 
$$
Z = \sum_s e^{-\beta E_s}.
$$ 
Now we have to evaluate $\partial \langle E \rangle / \partial \beta$. Substituting the Boltzmann distribution into the expression for the mean energy gives 
$$
\langle E \rangle = \frac{1}{Z}\sum_s E_s e^{-\beta E_s}.
$$ 
To evaluate the derivative with respect to $\beta$, we apply the quotient rule: 
$$
\frac{\partial \langle E \rangle}{\partial \beta} = -\frac{1}{Z}\sum_s E_s^2 e^{-\beta E_s} -\frac{1}{Z^2}\frac{\partial Z}{\partial \beta}\sum_s E_s e^{-\beta E_s}.
$$
From the definition of the partition function, we find 
$$
\frac{\partial Z}{\partial \beta} = -\sum_s E_s e^{-\beta E_s}.
$$
Substituting this result into the derivative of the mean energy yields 
$$
\frac{\partial \langle E \rangle}{\partial \beta} = -\frac{1}{Z}\sum_s E_s^2 e^{-\beta E_s} + \frac{1}{Z^2}\left(\sum_s E_s e^{-\beta E_s}\right)^2.
$$ 
Recognizing the expressions for the mean energy and the mean squared energy, we obtain 
$$
\frac{\partial \langle E \rangle}{\partial \beta} = -\langle E^2 \rangle + \langle E \rangle^2. 
$$
The energy variance is defined as 
$$
\text{Var}(E) = \langle E^2 \rangle - \langle E \rangle^2.
$$ 
Therefore, 
$$
\frac{\partial \langle E \rangle}{\partial \beta} = -\text{Var}(E).
$$ 
Substituting this result into the thermodynamic definition of the heat capacity gives 
$$
C_V = \frac{\text{Var}(E)}{k_B T^2}.
$$ 
For a system containing $N$ spins, the heat capacity per spin is defined as $c_V = C_V/N$. Hence, 
$$
\boxed{
c_V = \frac{\text{Var}(E)}{N k_B T^2}
}
$$


### Analytical Result for the Ising Chain

For the one-dimensional Ising chain with $J=1$, open boundary conditions, and no external magnetic field, the partition function is
$$
Z_N = 2(2\cosh\beta)^{N-1}, \qquad \beta = \frac{1}{k_B T}.
$$
The mean energy can be obtained directly from the partition function using
$$
\langle E\rangle = -\frac{\partial \ln Z_N}{\partial\beta}.
$$
Taking the logarithm of the partition function gives
$$
\ln Z_N = \ln 2 + (N-1)\ln(2\cosh\beta).
$$
Differentiating with respect to $\beta$, we obtain
$$
\langle E\rangle = -(N-1)\tanh\beta.
$$
With $\beta=1/k_B T$, the mean energy per spin becomes
$$
\boxed{
\frac{\langle E\rangle}{N} =-\frac{N-1}{N}\tanh\left(\frac{1}{k_B T}\right)
}
$$
The analytical heat capacity can then be obtained by differentiating the mean energy with respect to temperature:
$$
C_V = \frac{\partial\langle E\rangle}{\partial T}.
$$
Substituting the expression for the mean energy and applying the chain rule gives
$$
C_V = -(N-1)\frac{\partial}{\partial T}\tanh\left(\frac{1}{k_B T}\right) = \frac{N-1}{k_B T^2} \operatorname{sech}^2\left(\frac{1}{k_B T}\right).
$$
Therefore, the analytical heat capacity per spin is
$$
\boxed{
c_V = \frac{C_V}{N} = \frac{N-1}{N k_B T^2} \operatorname{sech}^2\left(\frac{1}{k_B T}\right)
}
$$
These analytical expressions provide reference curves for comparison with the Monte Carlo estimates. In particular, the heat capacity can be calculated either from the temperature derivative of the mean energy or from the energy fluctuations, as derived above.

## Magnetic Susceptibility per Spin

### General (used in Monte Carlo)

The magnetic susceptibility per spin is defined as the derivative of the mean magnetization per spin with respect to the external field, evaluated at zero field: 
$$
\chi = \left(\frac{\partial \langle m \rangle}{\partial h}\right)_{T,h=0},
$$ 
where $m=M/N$ is the magnetization per spin. Since $N$ is constant, this can be written as 
$$
\chi = \frac{1}{N}\left(\frac{\partial \langle M \rangle}{\partial h}\right)_{T,h=0}.
$$ 
The mean magnetization is defined as 
$$
\langle M \rangle = \sum_s M_s P_s,
$$ 
where $P_s$ is the probability of finding the system in configuration $s$. In the presence of an external field, the Boltzmann distribution is given by 
$$
P_s = \frac{e^{-\beta (E_s - hM_s)}}{Z},
$$ 
with the canonical partition function 
$$
Z = \sum_s e^{-\beta (E_s -hM_s)}.
$$ 
Substituting the Boltzmann distribution into the expression for the mean magnetization gives 
$$
\langle M \rangle
= \frac{1}{Z}\sum_s M_s e^{-\beta (E_s - hM_s)}.
$$ 
To evaluate the derivative with respect to $h$, we apply the quotient rule. Since $\partial (E_s - hM_s)/\partial h=-M_s$, differentiating the partition function gives 
$$
\frac{\partial Z}{\partial h} = \beta \sum_s M_s e^{-\beta (E_s - hM_s)}.
$$ 
Differentiating the mean magnetization therefore yields 
$$
\frac{\partial \langle M \rangle}{\partial h} = \frac{\beta}{Z}\sum_s M_s^2 e^{-\beta (E_s- hM_s)} - \frac{\beta}{Z^2}\left(\sum_s M_s e^{-\beta (E_s - hM_s)}\right)^2.
$$ 
Recognizing the expressions for the mean magnetization and the mean squared magnetization, we obtain 
$$
\frac{\partial \langle M \rangle}{\partial h} = \beta \left(\langle M^2 \rangle - \langle M \rangle^2\right).
$$ 
The variance of the magnetization is defined as 
$$
\text{Var}(M) = \langle M^2 \rangle - \langle M \rangle^2.
$$ 
Hence, the magnetic susceptibility per spin is 
$$
\chi = \frac{\beta}{N}\text{Var}(M).
$$ 
Using $\beta=1/(k_B T)$, this becomes 
$$
\boxed{
\chi = \frac{\text{Var}(M)}{N k_B T}
}
$$

### Analytical Result for the Ising Chain

At zero external field, the mean magnetization vanishes due to the symmetry between configurations with all spins reversed. Therefore, the magnetic susceptibility per spin is
$$
\chi = \frac{\langle M^2 \rangle}{N k_B T}.
$$
The total magnetization is $M=\sum_{i=1}^{N}s_i$, so its square is
$$
M^2 = \sum_{i=1}^{N}s_i^2 +2\sum_{i<j}s_i s_j.
$$
Since $s_i^2=1$, taking the thermal average gives
$$
\langle M^2\rangle = N+2\sum_{i<j}\langle s_i s_j\rangle.
$$
For a one-dimensional Ising chain with open boundary conditions and zero external field, the spin correlation function is
$$
\langle s_i s_j\rangle = \left[\tanh(\beta J)\right]^{|i-j|}.
$$
Grouping pairs of spins by their separation $r=|i-j|$, there are $N-r$ pairs at each separation. Hence,
$$
\boxed{
\langle M^2\rangle = N+2\sum_{r=1}^{N-1}(N-r)\left[\tanh(\beta J)\right]^r
}
$$
Substituting this into the expression for the susceptibility gives
$$
\boxed{
\chi = \frac{1}{N k_B T}\left[N + 2\sum_{r=1}^{N-1}(N-r)\left(\tanh\frac{J}{k_B T}\right)^r\right]
}
$$
This analytical expression can be compared with the Monte Carlo estimate of the susceptibility obtained from the magnetization fluctuations.
